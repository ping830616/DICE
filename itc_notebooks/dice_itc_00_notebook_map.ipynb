# DICE ITC Notebook Map

These notebooks split the full DICE analysis into smaller paper-focused units. The original `dice_results_analysis.ipynb` remains the master reference notebook.

Recommended order:
1. `dice_itc_01_run_and_setup.ipynb`
2. `dice_itc_02_core_results.ipynb`
3. `dice_itc_03_dse_and_complexity.ipynb`
4. `dice_itc_04_case_study_and_llm.ipynb`
5. `dice_itc_05_paper_bundle_and_repro.ipynb`

Use the first notebook to regenerate outputs. The later notebooks read the generated result files and focus on specific sections of the ITC paper.
